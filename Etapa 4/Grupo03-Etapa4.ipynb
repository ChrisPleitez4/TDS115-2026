7. Aplicar Chi cuadrado

```python
import numpy as np
from scipy import stats

def chi_square_test_safe(data, dist, params, bins=10):

    obs, edges = np.histogram(data, bins=bins)

    expected = []

    for i in range(len(edges)-1):
        p = dist.cdf(edges[i+1], *params) - dist.cdf(edges[i], *params)
        expected.append(len(data) * p)

    expected = np.array(expected)

    mask = expected >= 5

    obs = obs[mask]
    expected = expected[mask]

    if len(obs) < 2:
        return None, None

    # Ajustar para que ambas sumas coincidan
    expected = expected * (obs.sum() / expected.sum())

    chi2, p = stats.chisquare(obs, expected)

    return chi2, p


chi2_llegadas, p_llegadas = chi_square_test_safe(
    llegadas_seg,
    stats.expon,
    (0, np.mean(llegadas_seg))
)
chi2_servicios, p_servicios = chi_square_test_safe(
    servicios_seg,
    stats.expon,
    (0, np.mean(servicios_seg))
)
print("------------------PRUEBAS PARA EXPONENCIAL------------------")
print("1.LLEGADAS")
print("Chi²:", chi2_llegadas)
print("p-value:", p_llegadas)
alpha = 0.05
if p_llegadas > alpha:
    print("No se rechaza H0")
else:
    print("Se rechaza H0")
    
print("2.SERVICIOS")
print("Chi²:", chi2_servicios)
print("p-value:", p_servicios)
if p_servicios > alpha:
    print("No se rechaza H0")
else:
    print("Se rechaza H0")
```

```python

mu_llegadas = np.mean(llegadas_seg)
sigma_llegadas = np.std(llegadas_seg)

mu_servicios = np.mean(servicios_seg)
sigma_servicios = np.std(servicios_seg)

chi2_llegadas, p_llegadas = chi_square_test_safe(
    llegadas_seg,
    stats.norm,
    (mu_llegadas, sigma_llegadas)
)

chi2_servicios, p_servicios = chi_square_test_safe(
    servicios_seg,
    stats.norm,
    (mu_servicios, sigma_servicios)
)
print("------------------PRUEBAS PARA NORMAL------------------")
print("1.LLEGADAS")
print("Chi²:", chi2_llegadas)
print("p-value:", p_llegadas)
alpha = 0.05
if p_llegadas > alpha:
    print("No se rechaza H0")
else:
    print("Se rechaza H0")

print("2.SERVICIOS")
print("Chi²:", chi2_servicios)
print("p-value:", p_servicios)
if p_servicios > alpha:
    print("No se rechaza H0")
else:
    print("Se rechaza H0")
```

```python
shape_l, loc_l, scale_l = stats.gamma.fit(llegadas_seg)

shape_s, loc_s, scale_s = stats.gamma.fit(servicios_seg)

chi2_llegadas, p_llegadas = chi_square_test_safe(
    llegadas_seg,
    stats.gamma,
    (shape_l, loc_l, scale_l)
)

chi2_servicios, p_servicios = chi_square_test_safe(
    servicios_seg,
    stats.gamma,
    (shape_s, loc_s, scale_s)
)

print("------------------PRUEBAS PARA GAMMA------------------")

print("1.LLEGADAS")
print("Chi²:", chi2_llegadas)
print("p-value:", p_llegadas)

if p_llegadas > alpha:
    print("No se rechaza H0")
else:
    print("Se rechaza H0")

print("\n2.SERVICIOS")
print("Chi²:", chi2_servicios)
print("p-value:", p_servicios)

if p_servicios > alpha:
    print("No se rechaza H0")
else:
    print("Se rechaza H0")
```

```python
c_l, loc_l, scale_l = stats.weibull_min.fit(llegadas_seg)

c_s, loc_s, scale_s = stats.weibull_min.fit(servicios_seg)

chi2_llegadas, p_llegadas = chi_square_test_safe(
    llegadas_seg,
    stats.weibull_min,
    (c_l, loc_l, scale_l)
)

chi2_servicios, p_servicios = chi_square_test_safe(
    servicios_seg,
    stats.weibull_min,
    (c_s, loc_s, scale_s)
)

print("------------------PRUEBAS PARA WEIBULL------------------")

print("1.LLEGADAS")
print("Chi²:", chi2_llegadas)
print("p-value:", p_llegadas)

if p_llegadas > alpha:
    print("No se rechaza H0")
else:
    print("Se rechaza H0")

print("\n2.SERVICIOS")
print("Chi²:", chi2_servicios)
print("p-value:", p_servicios)

if p_servicios > alpha:
    print("No se rechaza H0")
else:
    print("Se rechaza H0")
```
